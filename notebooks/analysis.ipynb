```python
from pathlib import Path
import pandas as pd

results = Path('../data/results')
table_path = results / 'comparison_table.csv'
comparison = pd.read_csv(table_path) if table_path.exists() else pd.DataFrame()
comparison
```

```python
metrics = [column for column in ['HOTA', 'IDF1', 'MOTA', 'ID_switches', 'fps'] if column in comparison]
comparison[['run_name', *metrics]].sort_values('run_name') if not comparison.empty else 'Chưa có kết quả.'
```

Đối chiếu ít nhất ba seed GA và báo cáo mean ± std. Chỉ so sánh các run trên cùng video/dataset, cùng detector/checkpoint và cùng protocol đánh giá.
